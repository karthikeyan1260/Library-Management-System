<h1>Library Management System</h1>

```python
class Book:
    def __init__(self, Book_Name, Book_ID):
        self.Book_Name = Book_Name
        self.Book_ID = Book_ID


class LibraryManagement(Book):
    def __init__(self, Book_Name, Book_ID, Issued_Date, Period, Submission_Date):

        # Initialize Parent Class without super()
        Book.__init__(self, Book_Name, Book_ID)

        self.Issued_Date = Issued_Date
        self.Period = Period
        self.Submission_Date = Submission_Date
        self.Fine_Per_Day = 10

    def Convert_Date(self, Date):
        Day, Month, Year = map(int, Date.split("/"))
        return Day, Month, Year

    def is_leap_year(self, Year):
        return Year % 400 == 0 or (Year % 4 == 0 and Year % 100 != 0)

    def Days_in_Month(self, Month, Year):
        if Month == 2:
            return 29 if self.is_leap_year(Year) else 28
        elif Month in [4, 6, 9, 11]:
            return 30
        return 31

    def Total_Days(self, Date):
        Day, Month, Year = self.Convert_Date(Date)
        Days = Day

        for y in range(1, Year):
            Days += 366 if self.is_leap_year(y) else 365

        for m in range(1, Month):
            Days += self.Days_in_Month(m, Year)

        return Days

    def Calculate_Date(self, Total):
        Year = 1

        while True:
            Days = 366 if self.is_leap_year(Year) else 365
            if Total <= Days:
                break
            Total -= Days
            Year += 1

        Month = 1

        while True:
            Days = self.Days_in_Month(Month, Year)
            if Total <= Days:
                break
            Total -= Days
            Month += 1

        return f"{Total:02d}/{Month:02d}/{Year}"

    def Display_Details(self):
        Issued_Total = self.Total_Days(self.Issued_Date)
        Submission_Total = self.Total_Days(self.Submission_Date)

        Due_Total = Issued_Total + self.Period
        Due_Date = self.Calculate_Date(Due_Total)

        print("\n----- Library Management -----")
        print("Book Name        :", self.Book_Name)
        print("Book ID          :", self.Book_ID)
        print("Issued Date      :", self.Issued_Date)
        print("Period           :", self.Period, "Days")
        print("Due Date         :", Due_Date)
        print("Submission Date  :", self.Submission_Date)

        if Submission_Total <= Due_Total:
            print("Status           : Submitted On Time")
            print("Overdue Days     : 0")
            print("Total Fine       : ₹0")
        else:
            Overdue_Days = Submission_Total - Due_Total
            Fine = Overdue_Days * self.Fine_Per_Day

            print("Status           : Submitted Late")
            print("Overdue Days     :", Overdue_Days)
            print("Fine Per Day     : ₹", self.Fine_Per_Day)
            print("Total Fine       : ₹", Fine)


# Get User Input
Book_Name = input("Enter Book Name: ")
Book_ID = input("Enter Book ID: ")
Issued_Date = input("Enter Issued Date (DD/MM/YYYY): ")
Period = int(input("Enter Period (Days): "))
Submission_Date = input("Enter Submission Date (DD/MM/YYYY): ")

# Create Object
Book1 = LibraryManagement(
    Book_Name, Book_ID, Issued_Date, Period, Submission_Date
)

# Display Details
Book1.Display_Details()
```
